* Aggregation/agg() : ka use data ko summarize krne ke liye hota hai. Iski help se hum ek ya multiple columns par average, total, maximum, minimum aur count jaise calculations kar sakte hain.

* Ek column par aggregation : df["Rating"].agg("mean")

* groupby() ke saath agg() ka use :

df.groupby("Genre").agg({
    "Rating": "mean",
    "Runtime": "mean"
})

* Ek column par multiple aggregations :

df.groupby("Genre").agg({
    "Rating": ["mean", "max", "min"]
})

```python
# Table 1: Movie aur Genre
movies = csv_movies[["Series_Title", "Genre"]].copy()

# Table 2: Movie aur Rating
ratings = csv_movies[["Series_Title", "IMDB_Rating"]].copy()

# Movie name ko index banayein
movies = movies.set_index("Series_Title")
ratings = ratings.set_index("Series_Title")

# Dono tables join karein
result = movies.join(ratings)

print(result.head())
```

```python
#Left join
result=movies.join(ratings,how='left')
print(result)
```

```python

# Table 1: Movie details
movies = pd.DataFrame({
    "Genre": ["Drama", "Action", "Comedy", "Thriller"],
    "Year": [2016, 2023, 2009, 2022]
}, index=["Dangal", "Jawan", "3 Idiots", "Drishyam"])

# Table 2: Movie ratings
ratings = pd.DataFrame({
    "Rating": [8.4, 7.0, 8.4, 8.2],
    "Votes": [2000, 1500, 1800, 1200]
}, index=["Dangal", "Jawan", "PK", "Drishyam"])

print("MOVIES TABLE")
print(movies)

print("\nRATINGS TABLE")
print(ratings)
```
